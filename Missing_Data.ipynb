```python
#Missing data
```

```python
import numpy as np
import pandas as pd
```

```python
#find missing data
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
#Removing missing data
```

```python
df
```

```python
df.dropna()    #rows removed
```

```python
df.dropna(thresh=3)    #consider atleast 3 non-nan values
```

```python
df.fillna(0)
```

```python
df
```

```python
values= {'A':3,'B':3,'C':100,'D':200}
df.fillna(value= values)     #values assigned column wise
```

```python
df
```

```python
df.fillna(df.mean())
```
